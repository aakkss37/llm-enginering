```python
# Building a Web summarizer with Python and BeautifulSoup and LOCAL-LLM
```

```python
import requests
from bs4 import BeautifulSoup
from IPython.display import Markdown, display
```

```python
# Constants
OLLAMA__API_URL = "http://localhost:11434/api/chat"
HEADER = {"Content-Type": "application/json"}
MODEL = "qwen3:latest"
```

```python
# Create a messages list using the same format that we used for OpenAI

messages = [
    {"role": "user", "content": "Describe some of the business applications of Generative AI"}
]
```

```python
response = requests.post(
    OLLAMA__API_URL,
    headers=HEADER,
    json={"model": MODEL, "messages": messages},
)

import json

full_content = ""
for line in response.text.splitlines():
    if line.strip():
        obj = json.loads(line)
        # Each line contains a 'message' dict with 'content'
        content = obj.get("message", {}).get("content", "")
        full_content += content

print(full_content)
```

### ALTERNATIVELY WE CAN USE "`OLLAMA PACKAGE`" TO CALL OUR LOCAL LLM INSTEAD OF MAKING A `HTTP 'POST'` REQUEST

```python
import ollama
response = ollama.chat(
    model="qwen3:latest",
    messages=[
        {"role": "user", "content": "Describe some of the business applications of Generative AI"}
    ]
)
print(response['message']['content'])
```

## Alternative approach - using OpenAI python library to connect to Ollama

```python
from openai import OpenAI
ollama_via_openai = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')

response = ollama_via_openai.chat.completions.create(
    model=MODEL,
    messages=messages
)

print(response.choices[0].message.content)
```

## Streaming the response

```python
import ollama

stream = ollama.chat(
    model="qwen3:latest",
    messages=[
        {"role": "user", "content": "Describe some of the business applications of Generative AI"}
    ],
    stream=True
)

for chunk in stream:
    print(chunk['message']['content'], end="", flush=True)
```

```python
ollama_via_openai = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')

response_stream = ollama_via_openai.chat.completions.create(
    model=MODEL,
    messages=messages,
    stream=True
)

for chunk in response_stream:
    print(chunk.choices[0].delta.content, end="", flush=True)
```
