```python
import tiktoken

encoding = tiktoken.encoding_for_model("gpt-4.1-mini")

tokens = encoding.encode("Hi amar kumar, how are you doing today?")
```

```python
print(tokens)
```

```python
for token in tokens:
    token_text = encoding.decode([token])
    print(f"Token: {token}, Text: '{token_text}'")
```
